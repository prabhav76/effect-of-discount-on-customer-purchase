#### Audit Promotional Exposure Fields

Inspect the display, mailer and promotion-record fields in the enriched transaction dataset. Determine how promotional exposure is represented and whether the available data supports a meaningful comparison of purchasing behaviour.

```python
import pandas as pd
from collections import Counter

data_path = "../data/processed/transactions_enriched.csv"

promo_columns = [
    "display_values",
    "mailer_values",
    "promotion_records"
]

promo_counts = {col: Counter() for col in promo_columns}
missing_counts = {col: 0 for col in promo_columns}

total_rows = 0

for chunk in pd.read_csv(
    data_path,
    usecols=promo_columns,
    chunksize=250_000
):
    total_rows += len(chunk)

    for col in promo_columns:
        missing_counts[col] += chunk[col].isna().sum()
        promo_counts[col].update(
            chunk[col].dropna().astype(str).value_counts().to_dict()
        )

print("TOTAL TRANSACTION RECORDS:", total_rows)

for col in promo_columns:
    print("\n" + "=" * 65)
    print(col.upper())
    print("=" * 65)

    print("Missing records:", missing_counts[col])
    print("Non-missing records:", total_rows - missing_counts[col])
    print("Unique recorded values:", len(promo_counts[col]))

    print("\nMost frequent values:")
    display(
        pd.DataFrame(
            promo_counts[col].most_common(20),
            columns=["Value", "Record_Count"]
        )
    )
```

#### Examine Promotion Code Distributions

Summarise the display, mailer and promotion-record values, including their combinations and missingness. This establishes the available promotional exposure groups.

```python
import pandas as pd

data_path = "../data/processed/transactions_enriched.csv"

promotion_columns = [
    "display_values",
    "mailer_values",
    "promotion_records"
]

value_counts = {col: {} for col in promotion_columns}
combination_counts = {}
total_rows = 0

for chunk in pd.read_csv(
    data_path,
    usecols=promotion_columns,
    chunksize=250_000,
    low_memory=False
):
    total_rows += len(chunk)

    for col in promotion_columns:
        counts = chunk[col].value_counts(dropna=False)

        for value, count in counts.items():
            value_counts[col][value] = (
                value_counts[col].get(value, 0) + count
            )

    combinations = (
        chunk.fillna("MISSING")
        .astype(str)
        .value_counts()
    )

    for combination, count in combinations.items():
        combination_counts[combination] = (
            combination_counts.get(combination, 0) + count
        )

print("TOTAL RECORDS:", total_rows)

for col in promotion_columns:
    print(f"\n{col.upper()} DISTRIBUTION")

    display(
        pd.DataFrame(
            sorted(
                value_counts[col].items(),
                key=lambda x: (-x[1], str(x[0]))
            ),
            columns=["Value", "Record_Count"]
        )
    )

print("\nMOST COMMON PROMOTION COMBINATIONS")

display(
    pd.DataFrame(
        [
            list(combination) + [count]
            for combination, count in sorted(
                combination_counts.items(),
                key=lambda x: x[1],
                reverse=True
            )[:25]
        ],
        columns=promotion_columns + ["Record_Count"]
    )
)
```


#### Compare Transactions by Recorded Promotion Exposure

Compare sales, quantity and discount measures for records with and without recorded promotional metadata. Treat missing metadata as “not recorded,” not proof that no promotion was available.


```python

promo_path = "../data/processed/transactions_promo.csv"

summary = {
    "With recorded promotion metadata": {
        "records": 0, "sales": 0.0, "quantity": 0.0,
        "retail_discount": 0.0, "coupon_discount": 0.0,
        "coupon_match_discount": 0.0, "total_discount": 0.0,
        "discount_pct_sum": 0.0, "discount_pct_count": 0
    },
    "Without recorded promotion metadata": {
        "records": 0, "sales": 0.0, "quantity": 0.0,
        "retail_discount": 0.0, "coupon_discount": 0.0,
        "coupon_match_discount": 0.0, "total_discount": 0.0,
        "discount_pct_sum": 0.0, "discount_pct_count": 0
    }
}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    has_metadata = (
        chunk[["display_values", "mailer_values", "promotion_records"]]
        .notna()
        .any(axis=1)
    )

    chunk["promotion_group"] = has_metadata.map({
        True: "With recorded promotion metadata",
        False: "Without recorded promotion metadata"
    })

    for group_name, group in chunk.groupby("promotion_group"):
        s = summary[group_name]
        s["records"] += len(group)

        for column, key in [
            ("SALES_VALUE", "sales"),
            ("QUANTITY", "quantity"),
            ("retail_discount_amount", "retail_discount"),
            ("coupon_discount_amount", "coupon_discount"),
            ("coupon_match_discount_amount", "coupon_match_discount"),
            ("total_discount", "total_discount")
        ]:
            s[key] += group[column].sum()

        valid_pct = group["discount_percentage"].dropna()
        s["discount_pct_sum"] += valid_pct.sum()
        s["discount_pct_count"] += valid_pct.count()

results = []

for group_name, s in summary.items():
    n = s["records"]
    results.append({
        "Promotion_Group": group_name,
        "Records": n,
        "Record_Share_Percent": n / total_rows * 100,
        "Mean_Sales_Per_Record": s["sales"] / n,
        "Mean_Quantity_Per_Record": s["quantity"] / n,
        "Mean_Retail_Discount_Per_Record": s["retail_discount"] / n,
        "Mean_Coupon_Discount_Per_Record": s["coupon_discount"] / n,
        "Mean_Coupon_Match_Discount_Per_Record": s["coupon_match_discount"] / n,
        "Mean_Total_Discount_Per_Record": s["total_discount"] / n,
        "Mean_Discount_Percentage": (
            s["discount_pct_sum"] / s["discount_pct_count"]
            if s["discount_pct_count"] else np.nan
        ),
        "Valid_Discount_Percentage_Records": s["discount_pct_count"]
    })

promotion_comparison = pd.DataFrame(results)
display(promotion_comparison.round(4))
```


#### Compare Display and Mailer Promotion Records

Compare transaction sales and discount measures across display-only, mailer-only, combined display-and-mailer, and unrecorded promotion metadata groups.


```python

promotion_groups = {
    "Display and mailer recorded": {"records": 0, "sales": 0.0, "discount": 0.0, "pct_sum": 0.0, "pct_count": 0},
    "Display only": {"records": 0, "sales": 0.0, "discount": 0.0, "pct_sum": 0.0, "pct_count": 0},
    "Mailer only": {"records": 0, "sales": 0.0, "discount": 0.0, "pct_sum": 0.0, "pct_count": 0},
    "Neither recorded": {"records": 0, "sales": 0.0, "discount": 0.0, "pct_sum": 0.0, "pct_count": 0}
}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    has_display = chunk["display_values"].notna()
    has_mailer = chunk["mailer_values"].notna()

    conditions = [
        has_display & has_mailer,
        has_display & ~has_mailer,
        ~has_display & has_mailer
    ]
    choices = [
        "Display and mailer recorded",
        "Display only",
        "Mailer only"
    ]

    chunk["promotion_group"] = np.select(
        conditions, choices, default="Neither recorded"
    )

    for group_name, group in chunk.groupby("promotion_group"):
        s = promotion_groups[group_name]
        s["records"] += len(group)
        s["sales"] += group["SALES_VALUE"].sum()
        s["discount"] += group["total_discount"].sum()

        valid_pct = group["discount_percentage"].dropna()
        s["pct_sum"] += valid_pct.sum()
        s["pct_count"] += valid_pct.count()

promotion_type_results = []

for group_name, s in promotion_groups.items():
    n = s["records"]
    promotion_type_results.append({
        "Promotion_Type": group_name,
        "Records": n,
        "Record_Share_Percent": n / total_rows * 100,
        "Mean_Sales_Per_Record": s["sales"] / n if n else np.nan,
        "Mean_Total_Discount_Per_Record": s["discount"] / n if n else np.nan,
        "Mean_Discount_Percentage": (
            s["pct_sum"] / s["pct_count"] if s["pct_count"] else np.nan
        )
    })

promotion_type_summary = pd.DataFrame(promotion_type_results)
display(promotion_type_summary.round(4))
```


#### Step 12: Examine Coupon Discount Activity

Measure how many transaction records contain coupon discounts or coupon-match discounts, and quantify their contribution to total discounts.


```python

coupon_totals = {
    "records": 0,
    "coupon_discount_records": 0,
    "coupon_match_records": 0,
    "either_coupon_records": 0,
    "coupon_discount_sum": 0.0,
    "coupon_match_sum": 0.0,
    "total_discount_sum": 0.0
}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    coupon = chunk["COUPON_DISC"].fillna(0)
    match = chunk["COUPON_MATCH_DISC"].fillna(0)

    coupon_totals["records"] += len(chunk)
    coupon_totals["coupon_discount_records"] += (coupon != 0).sum()
    coupon_totals["coupon_match_records"] += (match != 0).sum()
    coupon_totals["either_coupon_records"] += (
        (coupon != 0) | (match != 0)
    ).sum()

    coupon_totals["coupon_discount_sum"] += coupon.sum()
    coupon_totals["coupon_match_sum"] += match.sum()
    coupon_totals["total_discount_sum"] += (
        chunk["total_discount"].fillna(0).sum()
    )

coupon_activity = pd.DataFrame([
    {
        "Metric": "Records with coupon discount",
        "Count": coupon_totals["coupon_discount_records"]
    },
    {
        "Metric": "Records with coupon-match discount",
        "Count": coupon_totals["coupon_match_records"]
    },
    {
        "Metric": "Records with either coupon discount",
        "Count": coupon_totals["either_coupon_records"]
    }
])

display(coupon_activity.assign(
    Record_Share_Percent=lambda x:
        x["Count"] / coupon_totals["records"] * 100
).round(4))

print("Total coupon discount:", round(coupon_totals["coupon_discount_sum"], 2))
print("Total coupon-match discount:", round(coupon_totals["coupon_match_sum"], 2))
print("Total discount across all records:", round(coupon_totals["total_discount_sum"], 2))
```


#### Step 13: Compare Coupon and Non-Coupon Transactions

Compare sales and discount measures between transaction records with and without coupon discounts. Quantity is excluded because its distribution is highly skewed.


```python

coupon_groups = {
    "Coupon discount recorded": {
        "records": 0, "sales": 0.0, "total_discount": 0.0,
        "coupon_discount": 0.0, "discount_pct_sum": 0.0,
        "discount_pct_count": 0
    },
    "No coupon discount recorded": {
        "records": 0, "sales": 0.0, "total_discount": 0.0,
        "coupon_discount": 0.0, "discount_pct_sum": 0.0,
        "discount_pct_count": 0
    }
}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    has_coupon = (
        chunk["COUPON_DISC"].fillna(0) != 0
    )

    chunk["coupon_group"] = np.where(
        has_coupon,
        "Coupon discount recorded",
        "No coupon discount recorded"
    )

    for group_name, group in chunk.groupby("coupon_group"):
        s = coupon_groups[group_name]
        s["records"] += len(group)
        s["sales"] += group["SALES_VALUE"].sum()
        s["total_discount"] += group["total_discount"].fillna(0).sum()
        s["coupon_discount"] += group["COUPON_DISC"].fillna(0).sum()

        valid_pct = group["discount_percentage"].dropna()
        s["discount_pct_sum"] += valid_pct.sum()
        s["discount_pct_count"] += valid_pct.count()

coupon_comparison = []

for group_name, s in coupon_groups.items():
    n = s["records"]
    coupon_comparison.append({
        "Coupon_Group": group_name,
        "Records": n,
        "Record_Share_Percent": n / total_rows * 100,
        "Mean_Sales_Per_Record": s["sales"] / n,
        "Mean_Total_Discount_Per_Record": s["total_discount"] / n,
        "Mean_Coupon_Discount_Per_Record": s["coupon_discount"] / n,
        "Mean_Discount_Percentage": (
            s["discount_pct_sum"] / s["discount_pct_count"]
            if s["discount_pct_count"] else np.nan
        )
    })

coupon_comparison = pd.DataFrame(coupon_comparison)
display(coupon_comparison.round(4))
```


#### Step 14: Compare Coupon-Using and Non-Coupon Households

Aggregate transaction measures at household level before testing whether households with coupon usage differ from households without coupon usage. This avoids treating every transaction as an independent observation.


```python

from scipy.stats import mannwhitneyu

household_coupon_stats = {}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    chunk["has_coupon"] = chunk["COUPON_DISC"].fillna(0) != 0

    for household, group in chunk.groupby("household_key"):
        if household not in household_coupon_stats:
            household_coupon_stats[household] = {
                "records": 0,
                "sales": 0.0,
                "discount_pct_sum": 0.0,
                "discount_pct_count": 0,
                "coupon_records": 0
            }

        s = household_coupon_stats[household]
        s["records"] += len(group)
        s["sales"] += group["SALES_VALUE"].sum()
        s["coupon_records"] += group["has_coupon"].sum()

        valid_pct = group["discount_percentage"].dropna()
        s["discount_pct_sum"] += valid_pct.sum()
        s["discount_pct_count"] += valid_pct.count()

household_coupon_df = pd.DataFrame([
    {
        "household_key": household,
        "Coupon_Group": (
            "Coupon user" if s["coupon_records"] > 0
            else "Non-coupon user"
        ),
        "Mean_Sales_Per_Record": s["sales"] / s["records"],
        "Mean_Discount_Percentage": (
            s["discount_pct_sum"] / s["discount_pct_count"]
            if s["discount_pct_count"] else np.nan
        )
    }
    for household, s in household_coupon_stats.items()
])

display(
    household_coupon_df.groupby("Coupon_Group").agg(
        Households=("household_key", "count"),
        Mean_Sales_Per_Record=("Mean_Sales_Per_Record", "mean"),
        Mean_Discount_Percentage=("Mean_Discount_Percentage", "mean")
    ).round(4)
)

for metric in ["Mean_Sales_Per_Record", "Mean_Discount_Percentage"]:
    coupon_users = household_coupon_df.loc[
        household_coupon_df["Coupon_Group"] == "Coupon user", metric
    ].dropna()

    non_coupon_users = household_coupon_df.loc[
        household_coupon_df["Coupon_Group"] == "Non-coupon user", metric
    ].dropna()

    statistic, p_value = mannwhitneyu(
        coupon_users, non_coupon_users, alternative="two-sided"
    )

    print(
        f"{metric}: U-statistic={statistic:.2f}, "
        f"p-value={p_value:.6g}"
    )
```


#### Step 15: Test Household-Level Differences by Promotion Exposure

Compare households with and without recorded promotion metadata using the Mann–Whitney U test. Test mean sales per transaction record and mean discount percentage, treating each household as one observation.


```python

promotion_household_stats = {}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    has_metadata = (
        chunk[["display_values", "mailer_values", "promotion_records"]]
        .notna()
        .any(axis=1)
    )

    chunk["has_promotion_metadata"] = has_metadata

    for household, group in chunk.groupby("household_key"):
        if household not in promotion_household_stats:
            promotion_household_stats[household] = {
                "records": 0,
                "sales": 0.0,
                "discount_pct_sum": 0.0,
                "discount_pct_count": 0,
                "promotion_records": 0
            }

        s = promotion_household_stats[household]
        s["records"] += len(group)
        s["sales"] += group["SALES_VALUE"].sum()
        s["promotion_records"] += group["has_promotion_metadata"].sum()

        valid_pct = group["discount_percentage"].dropna()
        s["discount_pct_sum"] += valid_pct.sum()
        s["discount_pct_count"] += valid_pct.count()

promotion_household_df = pd.DataFrame([
    {
        "household_key": household,
        "Promotion_Group": (
            "Promotion metadata recorded"
            if s["promotion_records"] > 0
            else "No promotion metadata recorded"
        ),
        "Mean_Sales_Per_Record": s["sales"] / s["records"],
        "Mean_Discount_Percentage": (
            s["discount_pct_sum"] / s["discount_pct_count"]
            if s["discount_pct_count"] else np.nan
        )
    }
    for household, s in promotion_household_stats.items()
])

display(
    promotion_household_df.groupby("Promotion_Group").agg(
        Households=("household_key", "count"),
        Mean_Sales_Per_Record=("Mean_Sales_Per_Record", "mean"),
        Mean_Discount_Percentage=("Mean_Discount_Percentage", "mean")
    ).round(4)
)

for metric in ["Mean_Sales_Per_Record", "Mean_Discount_Percentage"]:
    exposed = promotion_household_df.loc[
        promotion_household_df["Promotion_Group"] == "Promotion metadata recorded",
        metric
    ].dropna()

    unrecorded = promotion_household_df.loc[
        promotion_household_df["Promotion_Group"] == "No promotion metadata recorded",
        metric
    ].dropna()

    if len(exposed) > 0 and len(unrecorded) > 0:
        statistic, p_value = mannwhitneyu(
            exposed, unrecorded, alternative="two-sided"
        )
        print(
            f"{metric}: U-statistic={statistic:.2f}, "
            f"p-value={p_value:.6g}"
        )
    else:
        print(
            f"{metric}: Test not possible; "
            f"one group has no households."
        )
```


#### Step 16: Compare Baskets With and Without Recorded Promotion Metadata

Aggregate transaction records by basket and compare basket-level sales and discount percentages according to recorded promotion exposure. This checks whether the transaction-level pattern also appears at basket level.


```python

basket_stats = {}

for chunk in pd.read_csv(promo_path, chunksize=250_000, low_memory=False):
    chunk["has_promotion_metadata"] = (
        chunk[["display_values", "mailer_values", "promotion_records"]]
        .notna()
        .any(axis=1)
    )

    for basket, group in chunk.groupby("BASKET_ID"):
        if basket not in basket_stats:
            basket_stats[basket] = {
                "sales": 0.0,
                "discount_pct_sum": 0.0,
                "discount_pct_count": 0,
                "has_promotion": False
            }

        s = basket_stats[basket]
        s["sales"] += group["SALES_VALUE"].sum()
        s["has_promotion"] = (
            s["has_promotion"] or group["has_promotion_metadata"].any()
        )

        valid_pct = group["discount_percentage"].dropna()
        s["discount_pct_sum"] += valid_pct.sum()
        s["discount_pct_count"] += valid_pct.count()

basket_promotion_df = pd.DataFrame([
    {
        "BASKET_ID": basket,
        "Promotion_Group": (
            "Promotion metadata recorded" if s["has_promotion"]
            else "No promotion metadata recorded"
        ),
        "Basket_Sales": s["sales"],
        "Mean_Discount_Percentage": (
            s["discount_pct_sum"] / s["discount_pct_count"]
            if s["discount_pct_count"] else np.nan
        )
    }
    for basket, s in basket_stats.items()
])

display(
    basket_promotion_df.groupby("Promotion_Group").agg(
        Baskets=("BASKET_ID", "count"),
        Mean_Basket_Sales=("Basket_Sales", "mean"),
        Median_Basket_Sales=("Basket_Sales", "median"),
        Mean_Discount_Percentage=("Mean_Discount_Percentage", "mean")
    ).round(4)
)

for metric in ["Basket_Sales", "Mean_Discount_Percentage"]:
    exposed = basket_promotion_df.loc[
        basket_promotion_df["Promotion_Group"] == "Promotion metadata recorded",
        metric
    ].dropna()

    unrecorded = basket_promotion_df.loc[
        basket_promotion_df["Promotion_Group"] == "No promotion metadata recorded",
        metric
    ].dropna()

    if len(exposed) > 0 and len(unrecorded) > 0:
        statistic, p_value = mannwhitneyu(
            exposed, unrecorded, alternative="two-sided"
        )
        print(
            f"{metric}: U-statistic={statistic:.2f}, "
            f"p-value={p_value:.6g}"
        )
    else:
        print(f"{metric}: Test not possible; one group is empty.")
```


#### Step 17: Measure the Effect Size of Basket-Level Differences

Calculate rank-biserial correlation alongside the Mann–Whitney U results to understand the magnitude and direction of the basket-level differences, not just their statistical significance.


```python

from scipy.stats import mannwhitneyu

effect_size_results = []

for metric in ["Basket_Sales", "Mean_Discount_Percentage"]:
    exposed = basket_promotion_df.loc[
        basket_promotion_df["Promotion_Group"] == "Promotion metadata recorded",
        metric
    ].dropna()

    unrecorded = basket_promotion_df.loc[
        basket_promotion_df["Promotion_Group"] == "No promotion metadata recorded",
        metric
    ].dropna()

    u_stat, p_value = mannwhitneyu(
        exposed, unrecorded, alternative="two-sided"
    )

    rank_biserial = (
        2 * u_stat / (len(exposed) * len(unrecorded))
    ) - 1

    effect_size_results.append({
        "Metric": metric,
        "Promotion_Baskets": len(exposed),
        "Unrecorded_Baskets": len(unrecorded),
        "U_Statistic": u_stat,
        "P_Value": p_value,
        "Rank_Biserial_Correlation": rank_biserial
    })

effect_size_df = pd.DataFrame(effect_size_results)
display(effect_size_df.round(6))
```


#### Step 18: Consolidate Statistical Test Results

Combine the household-level coupon tests, household-level promotion tests and basket-level promotion tests into one results table. Include sample sizes, Mann–Whitney U statistics, p-values and rank-biserial effect sizes.


```python

statistical_results = []

def add_mannwhitney_result(
    analysis, metric, group1, group2, values1, values2
):
    values1 = values1.dropna()
    values2 = values2.dropna()

    u_stat, p_value = mannwhitneyu(
        values1, values2, alternative="two-sided"
    )

    rank_biserial = (
        2 * u_stat / (len(values1) * len(values2))
    ) - 1

    statistical_results.append({
        "Analysis": analysis,
        "Metric": metric,
        "Group_1": group1,
        "Group_2": group2,
        "N_Group_1": len(values1),
        "N_Group_2": len(values2),
        "U_Statistic": u_stat,
        "P_Value": p_value,
        "Rank_Biserial_Correlation": rank_biserial
    })


# Household-level coupon comparison
for metric in ["Mean_Sales_Per_Record", "Mean_Discount_Percentage"]:
    add_mannwhitney_result(
        "Household coupon comparison",
        metric,
        "Coupon users",
        "Non-coupon users",
        household_coupon_df.loc[
            household_coupon_df["Coupon_Group"] == "Coupon user", metric
        ],
        household_coupon_df.loc[
            household_coupon_df["Coupon_Group"] == "Non-coupon user", metric
        ]
    )


# Household-level promotion comparison
for metric in ["Mean_Sales_Per_Record", "Mean_Discount_Percentage"]:
    add_mannwhitney_result(
        "Household promotion comparison",
        metric,
        "Promotion metadata recorded",
        "No promotion metadata recorded",
        promotion_household_df.loc[
            promotion_household_df["Promotion_Group"]
            == "Promotion metadata recorded", metric
        ],
        promotion_household_df.loc[
            promotion_household_df["Promotion_Group"]
            == "No promotion metadata recorded", metric
        ]
    )


# Basket-level promotion comparison
for metric in ["Basket_Sales", "Mean_Discount_Percentage"]:
    add_mannwhitney_result(
        "Basket-level promotion comparison",
        metric,
        "Promotion metadata recorded",
        "No promotion metadata recorded",
        basket_promotion_df.loc[
            basket_promotion_df["Promotion_Group"]
            == "Promotion metadata recorded", metric
        ],
        basket_promotion_df.loc[
            basket_promotion_df["Promotion_Group"]
            == "No promotion metadata recorded", metric
        ]
    )


statistical_results_df = pd.DataFrame(statistical_results)

display(statistical_results_df.round(6))
```


#### Step 19: Apply Holm Correction to Statistical Tests

Apply the Holm–Bonferroni correction across the six Mann–Whitney tests to control the family-wise error rate and identify which findings remain statistically significant after correction.


```python

from statsmodels.stats.multitest import multipletests

statistical_results_df["Holm_Adjusted_P_Value"] = multipletests(
    statistical_results_df["P_Value"],
    method="holm"
)[1]

statistical_results_df["Significant_After_Correction"] = (
    statistical_results_df["Holm_Adjusted_P_Value"] < 0.05
)

display(
    statistical_results_df[
        [
            "Analysis", "Metric", "P_Value",
            "Holm_Adjusted_P_Value",
            "Rank_Biserial_Correlation",
            "Significant_After_Correction"
        ]
    ].round(8)
)
```
