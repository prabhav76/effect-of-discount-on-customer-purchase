#### Investigate Duplicate Promotional Records
We identify whether duplicate product-store-week records
are exact duplicates or contain different promotional information.

```python
causal = data["causal_data"]

keys = ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]

duplicates = causal[
    causal.duplicated(subset=keys, keep=False)
]

print("Duplicate records:", len(duplicates))
print("Exact duplicate rows:", duplicates.duplicated().sum())

non_key = [c for c in causal.columns if c not in keys]

variation = duplicates.groupby(keys)[non_key].nunique(dropna=False)

conflicting = variation.gt(1).any(axis=1)

print("Duplicate key groups:", len(variation))
print("Groups with different promotional values:", conflicting.sum())
print("Groups with identical promotional values:", (~conflicting).sum())
```

#### Identify Relevant Promotional Records
We identify the product-store-week combinations present
in transactions to limit unnecessary promotional data.

```python
transactions = data["transaction_data"]

transaction_keys = transactions[
    ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]
].drop_duplicates()

print("Unique transaction combinations:", len(transaction_keys))
```

#### Filter Relevant Promotional Records
We keep only promotional records matching product-store-week
combinations found in our transaction data.

```python
causal = data["causal_data"]

causal_relevant = causal.merge(
    transaction_keys,
    on=["PRODUCT_ID", "STORE_ID", "WEEK_NO"],
    how="inner"
)

print("Original causal rows:", len(causal))
print("Relevant causal rows:", len(causal_relevant))
print("Unique relevant combinations:",
      causal_relevant[["PRODUCT_ID", "STORE_ID", "WEEK_NO"]].drop_duplicates().shape[0])
```

#### Aggregate Promotional Information
We combine multiple promotional records for the same
product, store and week without losing distinct values.

```python
causal_agg = (
    causal_relevant.groupby(
        ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]
    )
    .agg(
        display_values=("display", lambda x: ",".join(sorted(x.unique()))),
        mailer_values=("mailer", lambda x: ",".join(sorted(x.unique()))),
        promotion_records=("display", "size")
    )
    .reset_index()
)

print("Aggregated rows:", len(causal_agg))
print("Duplicate keys:", causal_agg.duplicated(
    ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]
).sum())

causal_agg.head()
```

#### Validate Record Preservation

Check whether the aggregation preserved all relevant promotional records and whether each product-store-week combination is unique.

```python
print("Relevant causal rows:", len(causal_relevant))
print("Aggregated promotion records:", causal_agg["promotion_records"].sum())
print("Aggregated rows:", len(causal_agg))

print(
    "All records preserved:",
    causal_agg["promotion_records"].sum() == len(causal_relevant)
)

print(
    "Duplicate keys:",
    causal_agg.duplicated(
        ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]
    ).sum()
)
```

#### Check Missing Values and Promotion Counts

Check for missing values in the aggregated dataset and examine how many promotional records are associated with each product-store-week combination.

```python
print("Missing values:")
print(causal_agg.isnull().sum())

print("\nPromotion records per combination:")
print(causal_agg["promotion_records"].value_counts().sort_index())

print("\nMaximum promotional records for one combination:")
print(causal_agg["promotion_records"].max())
```

#### Merge Transaction and Promotional Data

Combine transaction records with the aggregated promotional data using PRODUCT_ID, STORE_ID, and WEEK_NO. A left join ensures that all transaction records are retained, including those without matching promotional information.

```python
transactions_promo = transactions.merge(
    causal_agg,
    on=["PRODUCT_ID", "STORE_ID", "WEEK_NO"],
    how="left",
    validate="many_to_one"
)

print("Original transaction rows:", len(transactions))
print("Merged transaction rows:", len(transactions_promo))
print("Rows preserved:", len(transactions) == len(transactions_promo))

transactions_promo.head()
```

#### Create Discount Features

Convert recorded discounts into positive amounts and calculate the total recorded discount for each transaction.

```python
transactions_promo["retail_discount_amount"] = -transactions_promo["RETAIL_DISC"]
transactions_promo["coupon_discount_amount"] = -transactions_promo["COUPON_DISC"]
transactions_promo["coupon_match_discount_amount"] = -transactions_promo["COUPON_MATCH_DISC"]

transactions_promo["total_discount"] = (
    transactions_promo["retail_discount_amount"]
    + transactions_promo["coupon_discount_amount"]
    + transactions_promo["coupon_match_discount_amount"]
)

transactions_promo[
    ["RETAIL_DISC", "retail_discount_amount",
     "COUPON_DISC", "coupon_discount_amount",
     "COUPON_MATCH_DISC", "coupon_match_discount_amount",
     "total_discount"]
].head()
```

#### Clean Discount Values

Replace negative zero values with regular zero for cleaner and consistent discount calculations.

```python
discount_amount_cols = [
    "retail_discount_amount",
    "coupon_discount_amount",
    "coupon_match_discount_amount"
]

for col in discount_amount_cols:
    transactions_promo[col] = transactions_promo[col].clip(lower=0)

transactions_promo["total_discount"] = (
    transactions_promo[discount_amount_cols].sum(axis=1)
)

transactions_promo[discount_amount_cols + ["total_discount"]].head()
```

#### Clean Discount Features

Recalculate discount amounts from the original columns and standardise zero values while preserving all recorded values.

```python
discount_map = {
    "retail_discount_amount": "RETAIL_DISC",
    "coupon_discount_amount": "COUPON_DISC",
    "coupon_match_discount_amount": "COUPON_MATCH_DISC"
}

for new_col, original_col in discount_map.items():
    transactions_promo[new_col] = -transactions_promo[original_col]
    transactions_promo.loc[transactions_promo[new_col] == 0, new_col] = 0.0

transactions_promo["total_discount"] = (
    transactions_promo[list(discount_map)].sum(axis=1)
)

transactions_promo[
    list(discount_map) + ["total_discount"]
].head()
```

#### Calculate Discount Percentage

Estimate the original price before discounts and calculate the percentage discount for each transaction.

```python
transactions_promo["estimated_original_value"] = (
    transactions_promo["SALES_VALUE"]
    + transactions_promo["total_discount"]
)

transactions_promo["discount_percentage"] = (
    transactions_promo["total_discount"]
    / transactions_promo["estimated_original_value"]
) * 100

transactions_promo[
    ["SALES_VALUE", "total_discount",
     "estimated_original_value", "discount_percentage"]
].head()
```

#### Validate Discount Percentages

Check for missing values, invalid percentages and unusual discount values before using this feature in analysis.

```python
print("Missing percentages:", transactions_promo["discount_percentage"].isna().sum())

print("Minimum:", transactions_promo["discount_percentage"].min())
print("Maximum:", transactions_promo["discount_percentage"].max())

print("\nDiscount percentage summary:")
print(transactions_promo["discount_percentage"].describe())

print("\nPercentages above 100:")
print((transactions_promo["discount_percentage"] > 100).sum())

print("Negative percentages:")
print((transactions_promo["discount_percentage"] < 0).sum())
```

#### Investigate Invalid Original Values

Check how many estimated original values are zero, negative or positive.

```python
print("Zero values:", (transactions_promo["estimated_original_value"] == 0).sum())
print("Negative values:", (transactions_promo["estimated_original_value"] < 0).sum())
print("Positive values:", (transactions_promo["estimated_original_value"] > 0).sum())

print("Infinite percentages:", 
      transactions_promo["discount_percentage"].isin([float("inf"), -float("inf")]).sum())
```

#### Recalculate Valid Discount Percentages

Calculate discount percentages only when the estimated original value is positive and the recorded discount is non-negative.

```python
import numpy as np

valid = (
    (transactions_promo["estimated_original_value"] > 0)
    & (transactions_promo["total_discount"] >= 0)
)

transactions_promo["discount_percentage"] = np.nan

transactions_promo.loc[valid, "discount_percentage"] = (
    transactions_promo.loc[valid, "total_discount"]
    / transactions_promo.loc[valid, "estimated_original_value"]
) * 100

print(transactions_promo["discount_percentage"].describe())
print("Missing percentages:", transactions_promo["discount_percentage"].isna().sum())
```

#### Investigate Invalid Discount Records

Identify why some transactions have missing discount percentages.

```python
original = transactions_promo["estimated_original_value"]
discount = transactions_promo["total_discount"]

print("Non-positive original value:", (original <= 0).sum())
print("Negative total discount:", (discount < 0).sum())
print("Missing original value:", original.isna().sum())
print("Missing total discount:", discount.isna().sum())

print("\nRecords with negative total discount:")
print(transactions_promo.loc[
    discount < 0,
    ["SALES_VALUE", "RETAIL_DISC", "total_discount",
     "estimated_original_value"]
].head())
```

#### Correct Discount Calculations

Remove insignificant floating-point noise and count only actual recorded discounts as positive discount amounts.

```python
# Round monetary values to two decimal places before calculation

transactions_promo["retail_discount_amount"] = (
    -transactions_promo["RETAIL_DISC"].round(2)
).clip(lower=0)

transactions_promo["coupon_discount_amount"] = (
    -transactions_promo["COUPON_DISC"].round(2)
).clip(lower=0)

transactions_promo["coupon_match_discount_amount"] = (
    -transactions_promo["COUPON_MATCH_DISC"].round(2)
).clip(lower=0)

discount_cols = [
    "retail_discount_amount",
    "coupon_discount_amount",
    "coupon_match_discount_amount"
]

transactions_promo["total_discount"] = (
    transactions_promo[discount_cols].sum(axis=1)
)

transactions_promo["estimated_original_value"] = (
    transactions_promo["SALES_VALUE"].round(2)
    + transactions_promo["total_discount"]
)

print("Negative total discounts:",
      (transactions_promo["total_discount"] < 0).sum())

print("Non-positive estimated original values:",
      (transactions_promo["estimated_original_value"] <= 0).sum())
```

#### Recalculate Discount Percentage

Calculate discount percentages only for transactions with a positive estimated original value.

```python
transactions_promo["discount_percentage"] = np.nan

valid = transactions_promo["estimated_original_value"] > 0

transactions_promo.loc[valid, "discount_percentage"] = (
    transactions_promo.loc[valid, "total_discount"]
    / transactions_promo.loc[valid, "estimated_original_value"]
) * 100

print(transactions_promo["discount_percentage"].describe())
print("\nMissing percentages:", transactions_promo["discount_percentage"].isna().sum())
print("Infinite percentages:", np.isinf(
    transactions_promo["discount_percentage"].dropna()
).sum())
```
